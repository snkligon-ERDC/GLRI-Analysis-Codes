```python
# ============================================================
# CELL 1 — Import libraries
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# make plots a little cleaner for teaching
plt.style.use("ggplot")

print("Libraries loaded successfully.")
```

```python
# ============================================================
# CELL 2 — Upload Excel workbook
# ============================================================

from google.colab import files

uploaded = files.upload()

# Get uploaded filename
WORKBOOK_PATH = list(uploaded.keys())[0]

print("Workbook uploaded:", WORKBOOK_PATH)
```

```python
# ============================================================
# CELL 3 — Inspect workbook sheets
# ============================================================

xls = pd.ExcelFile(WORKBOOK_PATH)

print("Sheets found in workbook:")
print(xls.sheet_names)
```

```python
# ============================================================
# CELL 4 — Load first sheet
# ============================================================

sheet_name = xls.sheet_names[0]

df = pd.read_excel(WORKBOOK_PATH, sheet_name=sheet_name)

# standardize column names
df.columns = [str(c).strip().lower() for c in df.columns]

print("Data shape:", df.shape)

df.head()
```

```python
# ============================================================
# CELL 5 — Identify numeric columns
# ============================================================

numeric_columns = df.select_dtypes(include=np.number).columns.tolist()

print("Numeric columns available for plotting:")
print(numeric_columns)
```

```python
# ============================================================
# CELL 6 — Summary statistics
# ============================================================

df[numeric_columns].describe()
```

```python
# ============================================================
# CELL 7 — Bar plot function
# ============================================================

def plot_bar_parameter(dataframe, parameter):

    if parameter not in dataframe.columns:
        raise ValueError(f"{parameter} not found in dataframe")

    plt.figure(figsize=(10,5))

    dataframe[parameter].plot(
        kind="bar",
        color="steelblue"
    )

    plt.title(f"Bar Plot of {parameter}")
    plt.xlabel("Sample Index")
    plt.ylabel(parameter)

    plt.tight_layout()
    plt.show()
```

```python
# ============================================================
# CELL 8 — Example plot
# ============================================================

parameter = numeric_columns[0]

print("Plotting:", parameter)

plot_bar_parameter(df, parameter)
```

```python
# ============================================================
# CELL 9 — Plot several parameters
# ============================================================

parameters_to_plot = numeric_columns[:4]

for param in parameters_to_plot:

    print("Plotting:", param)

    plot_bar_parameter(df, param)
```

```python
# ============================================================
# CELL 10 — Grouped bar plot by sample class
# ============================================================

if "sample_class" in df.columns:

    parameter = numeric_columns[0]

    summary = (
        df.groupby("sample_class")[parameter]
        .mean()
        .reset_index()
    )

    plt.figure(figsize=(6,4))

    plt.bar(summary["sample_class"], summary[parameter])

    plt.title(f"Mean {parameter} by Sample Class")
    plt.ylabel(parameter)
    plt.xlabel("Sample Class")

    plt.show()

else:
    print("No 'sample_class' column found.")
```

```python
# ============================================================
# CELL 11 — Export summary table
# ============================================================

summary_table = df[numeric_columns].mean().reset_index()

summary_table.columns = ["parameter", "mean_value"]

summary_table.to_excel("nutrient_summary.xlsx", index=False)

print("Summary saved as nutrient_summary.xlsx")
```
